```python
from pathlib import Path
import sqlite3
import pandas as pd

# Works whether the notebook's working folder is the project root or notebooks/
project_root = Path.cwd()
if not (project_root / "data" / "raw" / "stroke.db").exists():
    project_root = project_root.parent

db_path = project_root / "data" / "raw" / "stroke.db"
csv_path = project_root / "outputs" / "stroke_for_powerbi.csv"

with sqlite3.connect(db_path) as connection:
    stroke_data = pd.read_sql_query("SELECT * FROM stroke", connection)

stroke_data.to_csv(csv_path, index=False)
print(f"Exported {len(stroke_data)} rows to: {csv_path}")
```
