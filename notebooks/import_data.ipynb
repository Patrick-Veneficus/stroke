van kaggle gehaald en een pad gekregen naar de csv file toe:



```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("cavgarcia/new-stroke-prediction-dataset-nhanes-survey-data")

print("Path to dataset files:", path)
```

```python
from pathlib import Path

dataset_dir = Path(path)
csv_files = list(dataset_dir.rglob("*.csv"))
print(*csv_files, sep="\n")
```

in een pandas dataframe gezet:

```python
import pandas as pd

# read the csv
df = pd.read_csv(csv_files[0])

print(f"Loaded df: {df.shape[0]} rows, {df.shape[1]} columns")
```

```python
from pathlib import Path
import sqlite3

cwd = Path.cwd()
project_root = cwd.parent if cwd.name.lower() == "notebooks" else cwd
db_dir = project_root / "data" / "raw"
db_dir.mkdir(parents=True, exist_ok=True)

con = sqlite3.connect(db_dir / "stroke.db")
df.to_sql("stroke", con, if_exists="replace", index=False)
con.close()
```
